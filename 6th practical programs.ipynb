```
class BankAccount:
    def __init__(self, owner, balance=0):
        self.__owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Deposited ₹{amount}. Balance: ₹{self.__balance}")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
            print(f"Withdrawn ₹{amount}. Balance: ₹{self.__balance}")
        else:
            print("Insufficient funds!")

    def get_balance(self):
        return self.__balance

acc = BankAccount("Alice", 5000)
acc.deposit(2000)
acc.withdraw(1500)
print("Final Balance:", acc.get_balance())
```

```
import math

class Shape:
    def area(self):
        raise NotImplementedError("Subclass must implement area()")

    def describe(self):
        print(f"{self.__class__.__name__}: Area = {self.area():.2f}")

class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        return math.pi * self.r ** 2

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l, self.w = l, w
    def area(self):
        return self.l * self.w

class Triangle(Shape):
    def __init__(self, b, h):
        self.b, self.h = b, h
    def area(self):
        return 0.5 * self.b * self.h

shapes = [Circle(7), Rectangle(5, 10), Triangle(6, 8)]
for s in shapes:
    s.describe()
```

```
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks          # list of subject marks

    def display_details(self):
        print(f"Name: {self.name}, Roll No: {self.roll_no}")
        print("Marks:", self.marks)

    def average_marks(self):
        return sum(self.marks) / len(self.marks)

    def display_result(self):
        avg = self.average_marks()
        status = "Pass" if avg >= 40 else "Fail"   # threshold — verify with faculty
        print(f"Average: {avg:.2f} -> {status}")

s = Student("Kabir", 101, [85, 72, 91, 64])
s.display_details()
s.display_result()
```

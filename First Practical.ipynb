```
#fibonacci series in a loop
n = int(input("Enter the number of terms: "))

# First two terms
a, b = 0, 1

if n <= 0:
    print("Please enter a positive integer")
elif n == 1:
    print(f"Fibonacci sequence up to {n}: {a}")
else:
    print("Fibonacci sequence:")
    print(a, b, end=" ")
    for i in range(2, n):
        c = a + b
        print(c, end=" ")
        a, b = b, c
```

```
#Sum of natural numbers in for loop
n = int(input("Enter a positive integer: "))
sum_n = 0

# Loop from 1 to n (inclusive)
for i in range(1, n + 1):
    sum_n += i

print(f"Sum of first {n} natural numbers is: {sum_n}")
```

```
#Factorial numbers using for loop
n = int(input("Enter a number: "))

if n < 0:
    print("Factorial does not exist for negative numbers")
elif n == 0:
    print("Factorial of 0 is 1")
else:
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    print(f"Factorial of {n} is {factorial}")
```

```
#sum of digits using for loop
n = int(input("Enter a number: "))
temp = abs(n) # Using absolute value to handle negative inputs safely
sum_digits = 0
reverse_num = 0

while temp > 0:
    digit = temp % 10          # Extract the last digit
    sum_digits += digit        # Add to sum
    reverse_num = (reverse_num * 10) + digit # Append to reverse
    temp //= 10                # Remove the last digit

print(f"Sum of digits: {sum_digits}")
print(f"Reverse of number: {reverse_num}")
```

```
#Armstrong number using for loop
n = int(input("Enter a number: "))
temp = n
num_digits = len(str(n)) # Find the number of digits
sum_powers = 0

while temp > 0:
    digit = temp % 10
    sum_powers += digit ** num_digits
    temp //= 10

if sum_powers == n:
    print(f"{n} is an Armstrong number")
else:
    print(f"{n} is not an Armstrong number")
```

```
#Prime numbers using for loop
start = int(input("Enter the start of range: "))
end = int(input("Enter the end of range: "))

print(f"Prime numbers between {start} and {end} are:")

for num in range(start, end + 1):
    if num > 1: # Prime numbers are greater than 1
        is_prime = True
        # Check for factors up to the square root of the number
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break
        if is_prime:
            print(num, end=" ")
```

```
import numpy as np

scores = np.array([85, 72, 91, 64, 78, 88, 95, 70, 55, 82])
print("Scores:", scores)
print("Mean:", np.mean(scores))
print("Median:", np.median(scores))
print("Std Dev:", np.std(scores))
print("Min:", np.min(scores), "Max:", np.max(scores))
print("Variance:", np.var(scores))
print("Sorted:", np.sort(scores))
print("75th Percentile:", np.percentile(scores, 75))
```

```
import pandas as pd
import matplotlib.pyplot as plt

# Create DataFrame
data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Math": [85, 72, 91, 64, 78],
    "Science": [88, 65, 82, 90, 71],
    "English": [76, 80, 68, 74, 95]
}
df = pd.DataFrame(data)

# Analysis
df["Total"] = df[["Math", "Science", "English"]].sum(axis=1)
df["Average"] = df["Total"] / 3
print(df)
print("\nClass Average:")
print(df[["Math", "Science", "English"]].mean())

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Bar chart
df.plot(x="Name", y=["Math", "Science", "English"],
        kind="bar", ax=axes[0], title="Subject-wise Marks")

# Pie chart – average marks distribution
avg = df[["Math", "Science", "English"]].mean()
axes[1].pie(avg, labels=avg.index, autopct="%1.1f%%")
axes[1].set_title("Average Marks Distribution")

plt.tight_layout()
plt.savefig("analysis.png", dpi=150)   # save BEFORE show()
plt.show()
```

```
import numpy as np

marks = np.array([85, 72, 91, 64, 78, 88, 95, 70, 55, 82])
print("Marks:", marks)
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Maximum:", np.max(marks))
print("Minimum:", np.min(marks))
print("Standard Deviation:", np.std(marks))
print("Variance:", np.var(marks))
print("Sorted Marks:", np.sort(marks))
```

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Math": [85, 72, 91, 64, 78],
    "Science": [88, 65, 82, 90, 71],
    "English": [76, 80, 68, 74, 95]
})

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Bar chart – student-wise marks
df.plot(x="Name", y=["Math", "Science", "English"],
        kind="bar", ax=axes[0], title="Student-wise Marks")
axes[0].set_xlabel("Student")
axes[0].set_ylabel("Marks")
axes[0].legend()

# Line chart – marks comparison
df.plot(x="Name", y=["Math", "Science", "English"],
        marker="o", ax=axes[1], title="Marks Comparison")
axes[1].set_xlabel("Student")
axes[1].set_ylabel("Marks")
axes[1].legend()

# Pie chart – subject-wise average
avg = df[["Math", "Science", "English"]].mean()
axes[2].pie(avg, labels=avg.index, autopct="%1.1f%%")
axes[2].set_title("Subject-wise Average Marks")

plt.tight_layout()
plt.savefig("marks_analysis.png", dpi=150)   # save BEFORE show()
plt.show()
```

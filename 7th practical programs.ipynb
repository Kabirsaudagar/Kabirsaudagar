```
# Writing to file
with open("students.txt", "w") as f:
    f.write("101,Alice,85\n")
    f.write("102,Bob,72\n")
    f.write("103,Charlie,91\n")
print("Records written.")

# Reading from file
print("\nStudent Records:")
with open("students.txt", "r") as f:
    for line in f:
        roll, name, marks = line.strip().split(",")
        print(f"Roll: {roll}, Name: {name}, Marks: {marks}")
```

```
class InvalidAgeError(Exception):
    """Custom exception for invalid age"""
    pass

def verify_age(age):
    if age < 0:
        raise InvalidAgeError("Age cannot be negative.")
    elif age < 18:
        raise ValueError("Age must be 18 or above.")
    return True

try:
    age = int(input("Enter age: "))
    if verify_age(age):
        print("Access granted!")
except InvalidAgeError as e:
    print(f"Custom Error: {e}")
except ValueError as e:
    print(f"Value Error: {e}")
except Exception as e:
    print(f"Unexpected Error: {e}")
finally:
    print("Verification process complete.")
```

```
filename = input("Enter file name: ")
try:
    with open(filename, "r") as f:
        lines = f.readlines()
    content = "".join(lines)
    print("Number of lines:", len(lines))
    print("Number of words:", len(content.split()))
    print("Number of characters:", len(content))
except FileNotFoundError:
    print(f"'{filename}' not found.")
```

```
while input("Add a student? (y/n): ").strip().lower() == "y":
    roll = input("Roll No: ")
    name = input("Name: ")
    marks = input("Marks: ")
    with open("students.txt", "a") as f:      # append keeps old records
        f.write(f"{roll},{name},{marks}\n")

print("\nAll Records:")
with open("students.txt", "r") as f:
    for line in f:
        roll, name, marks = line.strip().split(",")
        print(f"Roll: {roll}, Name: {name}, Marks: {marks}")
```

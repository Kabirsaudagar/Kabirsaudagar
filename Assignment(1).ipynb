```
#Average of Two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
average = (num1 + num2) / 2
print(f"Average = {average}")
```

```
#Average of Three Numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))
average = (num1 + num2 + num3) / 3
print(f"Average = {average}")
```

```
#Square and Cube of a Number
num = float(input("Enter a number: "))
square = num ** 2
cube = num ** 3
print(f"Square = {square}")
print(f"Cube = {cube}")
```

```
#Area and Perimeter of a Square
side = float(input("Enter the side of the square: "))
area = side * side
perimeter = 4 * side
print(f"Area = {area}")
print(f"Perimeter = {perimeter}")
```

```
#Area and Perimeter of a Rectangle
length = float(input("Enter length: "))
breadth = float(input("Enter breadth: "))
area = length * breadth
perimeter = 2 * (length + breadth)
print(f"Area = {area}")
print(f"Perimeter = {perimeter}")
```

```
#Diameter and Circumference of a Circle
radius = float(input("Enter the radius: "))
diameter = 2 * radius
circumference = 2 * 3.14159 * radius
print(f"Diameter = {diameter}")
print(f"Circumference = {circumference}")
```

```
#Remainder and Quotient
num1 = float(input("Enter dividend: "))
num2 = float(input("Enter divisor: "))
quotient = num1 // num2
remainder = num1 % num2
print(f"Quotient = {quotient}")
print(f"Remainder = {remainder}")
```

```
#Half and Double of a Number
num = float(input("Enter a number: "))
half = num / 2
double = num * 2
print(f"Half = {half}")
print(f"Double = {double}")
```

```
#10%, 20%, and 50% of a Number
num = float(input("Enter a number: "))
ten_percent = num * 0.10
twenty_percent = num * 0.20
fifty_percent = num * 0.50
print(f"10% = {ten_percent}")
print(f"20% = {twenty_percent}")
print(f"50% = {fifty_percent}")
```

```
#Percentage of Marks Obtained
total_marks = float(input("Enter total marks: "))
obtained_marks = float(input("Enter marks obtained: "))
percentage = (obtained_marks / total_marks) * 100
print(f"Percentage = {percentage}%")
```

```
my_tuple = (15, 88, 3, 42, 79, 10)

max_value = max(my_tuple)
min_value = min(my_tuple)

print(f"Maximum value: {max_value}")
print(f"Minimum value: {min_value}")
```

```
numbers = []

for i in range(5):
    num = float(input(f"Enter number {i + 1}: "))
    numbers.append(num)

print("The stored list is:", numbers)
```

```
def count_words(sentence):
    # split() divides the string by whitespace into a list of words
    return len(sentence.strip().split())


# Example usage:
text = "Python programming is versatile and powerful."
print("Word count:", count_words(text))
```

```
# Formula: (Principal * Rate * Time) / 100
simple_interest = lambda p, r, t: (p * r * t) / 100

# Example usage (Principal = 1000, Rate = 5%, Time = 2 years):
interest = simple_interest(1000, 5, 2)
print("Simple Interest:", interest)
```

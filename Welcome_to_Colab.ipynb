```
area = lambda r: 3.14*r*r

rs = [1, 2, 3]
print(list(map(area, rs)))
```

```
def area(r):
    return 3.14 * r * r

print(area (10))
```

```
class Circle:
     def __init__(s, r):
         s.r = r
     def area(s):
         return 3.14*s.r*s.r

circle_instance = Circle(7)
print(circle_instance.area())
```

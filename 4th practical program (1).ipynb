```
#Program 1: Student Record System using Dictionary
students = {}
def add_student(roll, name, marks):
  students[roll] = {"name": name, "marks": marks}

def display_all():
 print(f"{"Roll":<8} {"Name":<15} {"Marks":<8} {"Grade"}")

 for roll, info in students.items():
  m = info["marks"]
  g = "Pass" if m >= 40 else "Fail"
  print(f"{roll:<8} {info['name']:<15} {m:<8} {g}")

add_student(101, "Alice", 85)
add_student(102, "Bob", 72)
add_student(103, "Charlie", 91)
display_all
```

```
#Program 2: Set Operations

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
print("Union:", A | B)
print("Intersection:", A & B)
print("Difference (A-B):", A - B)
print("Symmetric Difference:", A ^ B)
print("Is A subset of B?", A.issubset(B))
```
